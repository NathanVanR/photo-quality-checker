```python
import cv2
import numpy as np
import os
import random
```

```python
def create_bad_dataset(input_path, output_path):

    images = [f for f in os.listdir(input_path) if f.endswith(('.jpg', '.png', '.jpeg'))]

    for filename in images:
        img = cv2.imread(os.path.join(input_path, filename))

        failure_mode = random.choice(['blur', 'lighting', 'saturation'])

        if failure_mode == 'blur':
            processed = cv2.GaussianBlur(img, (15, 15), 0)
            label = "BLURRY"

        elif failure_mode == 'lighting':
            processed = np.clip(img.astype(float) * 0.2, 0, 255).astype(np.uint8)
            label = "DARK"

        elif failure_mode == 'saturation':
            hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
            h, s, v = cv2.split(hsv)
            s = np.clip(s.astype(float) * 0.1, 0, 255).astype(np.uint8)
            processed = cv2.cvtColor(cv2.merge([h, s, v]), cv2.COLOR_HSV2BGR)
            label = "DULL"

        new_name = f"bad_{label}_{filename}"
        cv2.imwrite(os.path.join(output_path, new_name), processed)
        print(f"Created {new_name}")

create_bad_dataset('../data/bad', '../data/bad_dataset')
```

```python
def get_stats(img_path):
    img = cv2.imread(img_path)
    if img is None: return None

    # 1. Lighting (Mean of Grayscale)
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    lighting = np.mean(gray)

    # 2. Blur (Variance of Laplacian)
    blur = cv2.Laplacian(gray, cv2.CV_64F).var()

    # 3. Saturation (Mean of S-channel in HSV)
    hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
    saturation = np.mean(hsv[:, :, 1])

    return lighting, blur, saturation

# Path to your folders
folders = ['data/good', 'data/bad']

print(f"{'Folder':<10} | {'File':<20} | {'Lighting':<10} | {'Blur':<10} | {'Saturation':<10}")
print("-" * 70)

for folder in folders:
    for file in os.listdir(folder):
        path = os.path.join(folder, file)
        stats = get_stats(path)
        if stats:
            l, b, s = stats
            print(f"{folder:<10} | {file[:20]:<20} | {l:10.2f} | {b:10.2f} | {s:10.2f}")
```
